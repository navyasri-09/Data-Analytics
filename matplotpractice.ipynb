```python
import matplotlib.pyplot as plt
import numpy as np
products=["Laptop","Mobile","Tablet","Headphone","Watch"]
sales=np.array([120,180,90,150,110])
profit=np.array([30,50,20,40,25])
print(products)
print(sales)
print(profit)
```

```python
#line chart
plt.plot(products,sales,marker="o")
plt.title("Product Sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(products,sales,marker="o",label="Sales")
plt.plot(products,profit,marker="o",label="Profits")
plt.title("Product Sales")
plt.xlabel("products")
plt.ylabel("sales and profit")
plt.legend()
plt.show()
```

```python
#bar chart
plt.bar(products,profit)
plt.title("Product profit")
plt.xlabel("Products")
plt.ylabel("Profit")
plt.show()
```

```python
#bar chart
plt.bar(sales,products)
plt.title("Product profit")
plt.xlabel("Products")
plt.ylabel("Profit")
plt.show()
```

```python
# horizontal bar chart
plt.barh(products,sales)
plt.title("Product profit")
plt.ylabel("Products")
plt.xlabel("Profit")
plt.show()
```

```python
plt.barh(profit,products,)  #(yaxis,xaxis)
plt.title("Product profit")
plt.xlabel("Products")
plt.ylabel("Profit")
plt.show()
```

```python
#grouped bar chart
x=np.arange(len(products))
width=0.35
plt.bar(x - width/2,sales,width,label="sales")
plt.bar(x + width/2,profit,width,label="profit")
plt.xticks(x,products)
plt.title("Sales Vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
x=np.arange(len(products))
width=0.35
plt.bar(x - width/2,sales,width,label="sales")
plt.bar(x - width/2,profit,width,label="profit")
plt.xticks(x,products)
plt.title("Sales Vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
#piechart
plt.pie(
    sales,
    labels = products,
    autopct= "%1.1f%%"
)
plt.title("product sales distribution")
plt.show()
```

```python
#donut chart
plt.pie(
    sales,
    labels = products,
    autopct= "%1.1f%%",
    wedgeprops={"width":0.4}
)
plt.title("product sales distribution")
plt.show()
```

```python
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F",
                "G", "H", "I", "J"],
    "Math": [65, 70, 75, 80, 85, 60, 90, 72,
             78, 88],
    "Python": [70, 75, 80, 85, 90, 65,
               95, 78, 82, 92],
    "SQL": [60, 68, 72, 78, 82, 55,
            88, 70, 75, 85]
}
df = pd.DataFrame(data)
print(df)
```

```python
#violin plot
import matplotlib.pyplot as plt
plt.violinplot(df["Math"])
plt.title("Distribution of Math Marks")
plt.ylabel("Marks")
plt.show()
```

```python
#stem plot
plt.stem(df["Student"], df["Math"])
plt.title("Student Math Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#conyour
import numpy as np
import matplotlib.pyplot as plt
# x = np.arange(1, 11)
# y = np.arange(1, 11)
x=[1,2,3]
y=[10,20]
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2
plt.contour(X, Y, Z)
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#heatmap
marks = df[["Math", "Python","SQL"]].values
plt.imshow(marks,cmap="Blues")
plt.title("Student Marks Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.colorbar()
```

```python
import pandas as pd

data ={
"Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Anjali", "Vikram", "Meena"],
"Age": [22, 24, None, 25, None, 26, 20, 27],
"City": ["Hyderabad", "Chennai", "Bangalore", "Hyderabad","Mumbai", None ,"Bangalore", "Mumbai"],
"X": [10,20,30,45,90,50,40,50],
"Y":[12,47,35,77,35,11,80,53],
"Z":[56,34,56,78,23,98,90,30]
}
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt


fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

ax.plot(df["X"],df["Y"],df["Z"])

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D salarys")

plt.show()
```

```python
import matplotlib.pyplot as plt


fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

ax.scatter(df["X"],df["Y"],df["Z"])

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D salarys")

plt.show()
```
