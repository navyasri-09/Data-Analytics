```python
#18
arr=np.array([10,20,30,40,50])
print(arr)
result=np.insert(arr,2,25)
print(result)
result2=np.delete(result,3)
print(result2)
print(np.sum(result))
print(np.min(result))
print(np.max(result))
print(np.mean(result))
```

```python
#19
marks=np.array([10,20,30,40,50])
print(np.var(marks))
print(np.std(marks))
print(np.percentile(marks,[25,75]))
print(np.argmax(marks))
print(np.argmin(marks))
print(np.log10(marks))
```

```python
#20
A=np.array([1,2,3])
B=np.array([4,5,6])
C=np.array([7,8,9])
print(np.concatenate((A,B)))
print("______________")
print(np.stack((A,B),axis=0))
print("______________")
print(np.stack((A,B),axis=1))
print("______________")
print(np.vstack((A,B,C)))
print("______________")
result=np.hstack((A,B,C))
print(result)
print("______________")
print(np.split(result,3))
```
