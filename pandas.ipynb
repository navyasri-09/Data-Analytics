```python
'''create a series of weather report  2025 12 months
jan 20
feb25
march none
april 30
may 40
june none
july none
aug 36
sep 22
oct 20
nov non
dec 10
'''
import pandas as pd
report={"Jan":20,"Feb":25,"March":None,"April":30,"May":40,"June":None,"July":None,"Aug":36,"Sep":22,"Oct":20,"Nov":None,"Dec":10}
s=pd.Series(report)
print(s)
```

```python
print(s.head())
print(s.tail())
print(len(s))
print(s.shape)
print(s.index)
print(s.values)
print(s["Oct"])
print(s[["Jan","March"]])
print(s.iloc[0])
print(s.iloc[0:9])
print(s>28)
print(s[s>28])
print(s.isna())
print(s.isna().sum())
print(s.fillna(s.sum()))
print(s.fillna(s.mean()))
print(s.fillna(s.min()))
print(s.fillna(s.max()))
print(s.fillna(s.count()))
print(s.fillna(s.median()))
print(s.fillna(s.std()))
```

```python
import pandas as pd
s1=pd.Series([100,200,300])
s2=pd.Series([10,20,30])
result=pd.concat([s1,s2])
print(result)
result=pd.concat([s1,s2],ignore_index=True)
print(result)
df=result.to_frame() #method to convert series into dataframe
print(df)
```

```python
import pandas as pd
s1=[0,2,4,6,8]
s=pd.Series(s1)
s2=[1,3,5,7,9]
t=pd.Series(s2)
s3=[2,3,5,7,11,13]
u=pd.Series(s3)
data={
    "Num":s,
    "Num1":t,
    "Num2":u
}
df=pd.DataFrame(data)
print(df)
df1=df.rename(columns={'Num':'a','Num1':'b','Num2':'c',},inplace=True)
print(df1)
print(df.iloc[0,1])
print(df.iloc[1,2])
print(df.iloc[3,1])
print(df[df['c']>10])
```

```python
import pandas as pd
data={
    "Name":["Ravi","Priya","Arun","Sneha","Kiran"],
    "Age":[22,24,21,25,23],
    "City":["Hyderabad","Chennai","Bangalore","Hyderabad","Mumbai"]
}
df=pd.DataFrame(data)
print(df)
print(df.head())
print(df.tail(2))
print(len(df))
print(df.shape[0]) #0-rows,1-columns
print(df.index)
print(df.values)
print(df.info()) #complete structure
print(df.describe()) #provides statistical info
print(df["Name"])
print(df[["Name","Age"]])
print(df.iloc[0])
print(df.iloc[0:3])
print(df.iloc[0,0])
print(df[df['Age']>23])
print(df[
    (df['Age']>23)&
    (df['City']=='Hyderabad')
])
#cleaning
print(df.isna())
print(df.isna().sum())
print(df.dropna())
```

```python
#practice question
import pandas as pd
data={
    "Name":["Ravi","Priya","Arun","Sneha","Kiran","Anjali","Vikram","Meena"],
    "Age":[22,24,21,25,23,26,20,27],
    "City":["Hyderabad","Chennai","Bangalore","Hyderabad","Mumbai","Chennai","Bangalore","Mumbai"],
    "Salary":[30000,40000,28000,50000,35000,45000,25000,55000]
}
df=pd.DataFrame(data)
print(df) #1
print(df['Name']) #2
print(df[['Name','Salary']]) #3
print(df[df['Age']>23]) #4
print(df[df['Salary']>40000]) #5
print(df[df['City']=='Chennai']) #6
#7
print(df[
    (df['Age']>20)&
    (df['Salary']>35000)
])
#8
print(df[
    (df['City']=='Mumbai')|
    (df['City']=='Bangalore')
    ])
#9
print(df[df["Salary"]>30000][['Name','Salary']])
#10
print(df.iloc[2:5])
#11
print(df.iloc[0:4][['Name','Age']])
#12
print(df.iloc[0,3])
```
