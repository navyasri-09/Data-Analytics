```python
#stack()--dimensions change
#1D
arr1=np.array([10,20,30])
arr2=np.array([40,50,60])
print(arr1)
print(arr2)
print(np.stack((arr1,arr2),axis=1))
print(np.stack((arr1,arr2),axis=0))
print(np.concatenate((arr1,arr2),axis=0))
```

```python

#2D
arr1=np.arange(10,20).reshape(2,5)
arr2=np.arange(1,11).reshape(2,5)
print(arr1)
print(arr2)
print("---------")
print(np.stack((arr1,arr2),axis=0))
print("-----------")
print(np.stack((arr1,arr2),axis=1))
```

```python
#3D
arr2=np.arange(1,31).reshape(2,3,5)
arr3=np.arange(21,51).reshape(2,3,5)
print(arr2)
print(arr3)
print(np.stack((arr2,arr3),axis=0))
print("----------")
print(np.stack((arr2,arr3),axis=1))
```

```python

#vertical stacking
#1D
arr1=np.array([10,20,30])
arr2=np.array([40,50,60])
print(arr1)
print(arr2)
print(np.vstack((arr1,arr2)))
print("---------------")
#2D
arr1=np.arange(10,20).reshape(2,5)
arr2=np.arange(1,11).reshape(2,5)
print(arr1)
print(arr2)
print(np.vstack((arr1,arr2)))
print("-----------------")
#3D
arr2=np.arange(1,31).reshape(2,3,5)
arr3=np.arange(21,51).reshape(2,3,5)
print(arr2)
print(arr3)
print(np.vstack((arr2,arr3)))
```

```python
#horizontal stacking
#1D
arr1=np.array([10,20,30])
arr2=np.array([40,50,60])
print(arr1)
print(arr2)
print(np.hstack((arr1,arr2)))
print("---------------")
#2D
arr1=np.arange(10,20).reshape(2,5)
arr2=np.arange(1,11).reshape(2,5)
print(arr1)
print(arr2)
print(np.hstack((arr1,arr2)))
print("-----------------")
#3D
arr2=np.arange(1,31).reshape(2,3,5)
arr3=np.arange(21,51).reshape(2,3,5)
print(arr2)
print(arr3)
print(np.hstack((arr2,arr3)))
```

```python
#vstack()
arr1=np.array([10,20,30])
arr2=np.array([40,50,60])
arr3=np.array([70,80,90])
print(arr1)
print(arr2)
print(arr3)
print(np.vstack((arr1,arr2,arr3)))
print("--------------")
#hstack()
print(np.hstack((arr1,arr2,arr3)))
```

```python
#split()
import numpy as np
arr=np.array([10,20,30,40,50,60,80,90,20])
print(np.split(arr,3))   #equal split
print(np.split(arr,[3,5]))  #position based
print(np.split(arr,[0,3,5]))
```
