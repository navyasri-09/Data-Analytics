```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
#load data
df=sns.load_dataset('titanic')
print(df)
```

```python
#lineplot
sns.lineplot(data=df,x='pclass',y='fare',hue='sex',marker='*')
```

```python
#barplot
sns.barplot(data=df,x='pclass',y='fare',hue='sex')
```

```python
corr=df[['age','fare','sibsp','parch']].corr()
sns.heatmap(
    corr,
    annot=True
)
plt.show()
```

```python
pivot=pd.pivot_table(
    df,values='survived',
    index='sex',
    columns='class',
    aggfunc='mean'
)
print(pivot)
```

```python
#visualize
sns.heatmap(
    pivot,annot=True,fmt='.2f'
)
```

```python
pivot=pd.pivot_table(
    df,values='survived',
    index=['sex','class'],
    aggfunc='mean'
)
print(pivot)
sns.heatmap(
    pivot,
    annot=True,
    fmt='.2f'
)
```

```python
#regression plot
sns.regplot(
    data=df,
    x='age',
    y='fare'
)
plt.show()
```

```python
#count plot
sns.countplot(
    data=df,
    x='class'
)
plt.show()
```

```python
#violin plot
sns.violinplot(
    data=df,
    x='class',
    y='age',
    hue='sex'
)
plt.title('Age Distribution by Class ans Sex')
plt.show()
```

```python
#density plot
sns.kdeplot(
    data=df,
    x='age',
    hue='sex',
    fill=True
)
plt.title("age density by gender")
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x='pclass',
    fill=True
)
plt.title("age density by gender")
plt.show()
```

```python
#pair plot
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.pairplot(
    df[['age','fare','sibsp']]
)
plt.show()
```

```python
g=sns.FacetGrid(
    df,
    col='sex'
)
g.map(
    sns.histplot,
    'age'
)
plt.show()
```

```python
g=sns.FacetGrid(
    df,
    col='sex'
)
g.map(
    sns.barplot,
    'age'
)
plt.show()
```

```python
g=sns.FacetGrid(
    df,
    col='sex'
)
g.map(
    sns.scatterplot,
    'age','fare'
)
plt.show()
```

```python
#joint plot--is a combination of relationship and distribution(Scatter and histogram)
sns.jointplot(
    df,
    x='age',
    y='fare'
)
plt.show()
```
